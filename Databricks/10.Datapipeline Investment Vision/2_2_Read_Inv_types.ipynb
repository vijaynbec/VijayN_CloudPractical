### Read file from volume and load into Raw table
- Set to Current Schema.
- Read files from Volume - Write into Table
- Validate the table

```python
%sql use schema investment_vision; select current_catalog(), current_schema(); 
```

```python
%python
raw_df = spark.read.format('csv').options(
    header=False,
    inferSchema=True,
    multiLine=True,
    quote='"',
    sep=','
).load(
    '/Volumes/workspace/investment_vision/files_volume/Types.txt'
)

# Rename columns from default _c0, _c1, _c2 to CUSIP, Sector, Industry_Type
raw_df = raw_df.toDF("CUSIP", "Sector", "Industry_Type")

display(raw_df)

raw_df.write.mode('overwrite').option('mergeSchema', 'true').saveAsTable(
    'workspace.investment_vision.raw_invest_types'
)
```
