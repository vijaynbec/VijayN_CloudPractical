```python
%sql
create or replace table investment_vision.gold_present_values_bycat AS
SELECT
    Invest_Catrgory,
    Present_Status,
    SUM(COALESCE(Sell, 0)) AS Total_Sell,
    SUM(COALESCE(Buy, 0)) AS Total_Buy,
    SUM(COALESCE(CDIV, 0)) AS Total_CDIV,
    SUM(COALESCE(current_market_value, 0)) AS Current_market_value,
    ROUND(
        SUM(
            CASE
                WHEN current_market_value IS NULL THEN COALESCE(Sell, 0) + COALESCE(CDIV, 0) - COALESCE(Buy, 0)
                ELSE COALESCE(Buy, 0) - COALESCE(Sell, 0) - COALESCE(CDIV, 0)
            END
        ),
        2
    ) AS `Current_Investment_Buy-Sell-CDIV`,
    round(
        SUM(
            CASE
                WHEN current_market_value IS NULL THEN COALESCE(Sell, 0) + COALESCE(CDIV, 0) - COALESCE(Buy, 0)
                ELSE COALESCE(current_market_value, 0) - (
                    CASE
                        WHEN current_market_value IS NULL THEN COALESCE(Sell, 0) + COALESCE(CDIV, 0) - COALESCE(Buy, 0)
                        ELSE COALESCE(Buy, 0) - COALESCE(Sell, 0) - COALESCE(CDIV, 0)
                    END
                )
            END
        ),
        2
    ) as `Current_Gain_Loss`,
    ROUND(
        CASE
            WHEN SUM(COALESCE(current_market_value, 0)) IS NULL OR SUM(COALESCE(current_market_value, 0)) = 0 THEN NULL
            ELSE (
                round(
                    SUM(
                        CASE
                            WHEN current_market_value IS NULL THEN COALESCE(Sell, 0) + COALESCE(CDIV, 0) - COALESCE(Buy, 0)
                            ELSE COALESCE(current_market_value, 0) - (
                                CASE
                                    WHEN current_market_value IS NULL THEN COALESCE(Sell, 0) + COALESCE(CDIV, 0) - COALESCE(Buy, 0)
                                    ELSE COALESCE(Buy, 0) - COALESCE(Sell, 0) - COALESCE(CDIV, 0)
                                END
                            )
                        END
                    ),
                    2
                ) / SUM(COALESCE(current_market_value, 0)) * 100
            )
        END, 2
    ) as `Current_Gain_Percent`
FROM investment_vision.current_details
GROUP BY Invest_Catrgory, Present_Status
```

```python
%sql
create or replace table investment_vision.gold_present_values_byIndtype as 
SELECT
    Industry_Type,
    SUM(COALESCE(Sell, 0)) AS Total_Sell,
    SUM(COALESCE(Buy, 0)) AS Total_Buy,
    SUM(COALESCE(CDIV, 0)) AS Total_CDIV,
    SUM(COALESCE(current_market_value, 0)) AS Current_market_value,
    ROUND(
        SUM(
            CASE
                WHEN current_market_value IS NULL THEN COALESCE(Sell, 0) + COALESCE(CDIV, 0) - COALESCE(Buy, 0)
                ELSE COALESCE(Buy, 0) - COALESCE(Sell, 0) - COALESCE(CDIV, 0)
            END
        ),
        2
    ) AS `Current_Investment_Buy-Sell-CDIV`,
    round(
        SUM(
            CASE
                WHEN current_market_value IS NULL THEN COALESCE(Sell, 0) + COALESCE(CDIV, 0) - COALESCE(Buy, 0)
                ELSE COALESCE(current_market_value, 0) - (
                    CASE
                        WHEN current_market_value IS NULL THEN COALESCE(Sell, 0) + COALESCE(CDIV, 0) - COALESCE(Buy, 0)
                        ELSE COALESCE(Buy, 0) - COALESCE(Sell, 0) - COALESCE(CDIV, 0)
                    END
                )
            END
        ),
        2
    ) as `Current_Gain_Loss`,
    ROUND(
        CASE
            WHEN SUM(COALESCE(current_market_value, 0)) IS NULL OR SUM(COALESCE(current_market_value, 0)) = 0 THEN NULL
            ELSE (
                round(
                    SUM(
                        CASE
                            WHEN current_market_value IS NULL THEN COALESCE(Sell, 0) + COALESCE(CDIV, 0) - COALESCE(Buy, 0)
                            ELSE COALESCE(current_market_value, 0) - (
                                CASE
                                    WHEN current_market_value IS NULL THEN COALESCE(Sell, 0) + COALESCE(CDIV, 0) - COALESCE(Buy, 0)
                                    ELSE COALESCE(Buy, 0) - COALESCE(Sell, 0) - COALESCE(CDIV, 0)
                                END
                            )
                        END
                    ),
                    2
                ) / SUM(COALESCE(current_market_value, 0)) * 100
            )
        END, 2
    ) as `Current_Gain_Percent`
FROM investment_vision.current_details
GROUP BY Industry_Type
ORDER BY `Current_Gain_Percent` DESC
```
