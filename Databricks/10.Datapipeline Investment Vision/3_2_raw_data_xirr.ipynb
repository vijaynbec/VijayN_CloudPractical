```python
%sql
CREATE OR REPLACE table workspace.investment_vision.raw_data_xirr AS
SELECT
    Settle_Date AS date,
    Instrument,
    CAST(CASE WHEN Trans_Code = 'Buy' THEN -Transaction_Amount_Clean ELSE 0 END AS DOUBLE) AS buy_amount,
    CAST(CASE WHEN Trans_Code = 'Sell' THEN Transaction_Amount_Clean ELSE 0 END AS DOUBLE) AS sell_amount,
    -- CAST(CASE WHEN Trans_Code = 'CDIV' THEN Transaction_Amount_Clean ELSE 0 END AS DOUBLE) AS cdiv_amount,
    CAST(0 AS DOUBLE) AS market_value
FROM workspace.investment_vision.valid_transactions
UNION ALL
SELECT
    current_date AS date,
    ticker AS Instrument,
    CAST(0 AS DOUBLE) AS buy_amount,
    CAST(0 AS DOUBLE) AS sell_amount,
  --  CAST(0 AS DOUBLE) AS cdiv_amount,
    CAST(MAX(current_market_value) AS DOUBLE) AS market_value
FROM workspace.investment_vision.current_portfolio_value
GROUP BY ticker
```
