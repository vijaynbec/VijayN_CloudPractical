```python
%sql
CREATE OR REPLACE TABLE workspace.investment_vision.gold_xirr_details AS
SELECT
  a.*,
  b.Sector AS Sector,
  b.Industry_Type AS Industry_Type,
  CASE 
    WHEN b.Sector LIKE '%ETF%' THEN 'ETF'
    ELSE 'Equity'
  END AS Invest_Catrgory_new,
  current_date() AS Current_Present_Date
FROM 
  workspace.investment_vision.silver_xirr_details a
LEFT JOIN 
  workspace.investment_vision.raw_invest_types b
ON 
  a.Instrument = b.CUSIP;
```

```python
%sql
select * from workspace.investment_vision.gold_xirr_details where Holding_Period_days > 200 and Market_Value > 0;
```
